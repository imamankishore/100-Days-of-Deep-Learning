```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
```

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("shenba/time-series-datasets")

print("Path to dataset files:", path)
```

```python
df = pd.read_csv('sales-of-shampoo-over-a-three-ye.csv')
```

```python
df.head()
```

```python
df = df.drop(columns=['humidity' , 'wind_speed' , 'meanpressure'])
```

```python
df= df.sample(1000)
```

```python
plt.scatter(df['Month'] , df['Sales of shampoo over a three year period'] , color='red')
```

```python
x1= df['Sales of shampoo over a three year period'].ewm(alpha=0.9).mean()
```

```python
x1
```

```python
df['ewm'] = x1
```

```python
df.head()
```

```python
plt.scatter(df['Month'] , df['Sales of shampoo over a three year period'] , color='red')
plt.plot(df['Month'] , x1, color='black' , label='alpha = 0.9')
plt.legend()
plt.show()
```
